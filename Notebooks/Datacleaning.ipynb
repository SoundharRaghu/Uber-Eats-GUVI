```python
import pandas as pd 
import numpy as np
df=pd.read_csv(r'C:\Users\Indhira Durai\OneDrive\Documents\GitHub\Uber-Eats-GUVI\Data\Raw\Uber_Eats_data.csv')
df.head(2)
```

```python
print(f"shapes:{df.shape}")
df.info()
```

```python
print(df['rate'].unique())
print(df['approx_cost(for two people)'].unique())
```

```python
df['rate'] = df['rate'].astype(str).str.replace('/5','').str.strip()
df['rate'] = df['rate'].replace('NEW',np.nan)
df['rate'] = pd.to_numeric(df['rate'],errors='coerce')
df['rate'].describe()
```

```python
print(df['rate'].unique())
```

```python
df.rename(columns={'approx_cost(for two people)':'cost'},inplace=True)
df['cost']=df['cost'].astype(str).str.replace(',','')
df['cost']=pd.to_numeric(df['cost'],errors='coerce')
df['cost'].describe()
```

```python
print(df['cost'].unique())
```

```python
before = len(df)
df.dropna(subset=['rate','cost','location','cuisines'], inplace=True)
after = len(df)
print(f"Before:{before}")
print(f"after:{after}")
print(f"Dropped:{before-after}")
```

```python
before = len(df)
df.drop_duplicates(inplace=True)
after =  len(df)
print(f"Before:{before}")
print(f"After:{after}")
print(f"Dropped:{before-after}")
```

```python
df.rename(columns={'listed_in(type)':'listed_type','listed_in(city)':'listed_city'}, inplace=True)
print(df.columns.tolist())
```

```python
df['price_segment']=pd.cut(df['cost'],bins=[0,300,700,100000],labels=['Low','Medium','Premium'])
```

```python
df['price_segment'].unique()
```

```python
def rate_cat(r):
    if r >= 4.0:
        return 'Excellent'
    elif r >= 3.0:
        return 'Good'
    else:
        return 'Poor'
    
df['rating_category']=df['rate'].apply(rate_cat)
print(df['rating_category'].value_counts())
```

```python
print(df['rating_category'].dtype)
```

```python
print(f"final shape: {df.shape}")
print(df.dtypes)
df.head()
```
