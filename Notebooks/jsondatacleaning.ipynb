```python
import pandas as pd
import numpy as np
import json
```

```python
import json
with open('../data/raw/orders.json') as f:orders=json.load(f)
jdf=pd.json_normalize(orders)
jdf.head()
```

```python
print(jdf.dtypes)
```

```python
jdf['order_date']=pd.to_datetime(jdf['order_date'])
jdf['year']=jdf['order_date'].dt.year
jdf['month']=jdf['order_date'].dt.month
jdf['month_name']=jdf['order_date'].dt.month_name()
jdf.head()
```
